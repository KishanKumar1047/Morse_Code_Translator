```python
import morse_talk as mtalk
import difflib

# Dataset of valid words (can be expanded)
DATASET = ["HELLO", "HELP", "WORLD", "HOUSE", "HERO", "HEY", "HILL", "HAPPY"]

# Morse dictionary for validation
MORSE_DICT = set([
    ".-", "-...", "-.-.", "-..", ".", "..-.", "--.", "....", "..", ".---",
    "-.-", ".-..", "--", "-.", "---", ".--.", "--.-", ".-.", "...", "-",
    "..-", "...-", ".--", "-..-", "-.--", "--..",
    ".----", "..---", "...--", "....-", ".....",
    "-....", "--...", "---..", "----.", "-----"
])

def recommend_words(word, dataset, n=5):
    """Suggest up to n similar words from dataset with Morse codes"""
    if all(ch in ".- " for ch in word.strip()):  
        # For invalid Morse, just suggest dataset words
        matches = dataset[:n]
    else:
        # For English words, find close matches
        matches = difflib.get_close_matches(word.upper(), dataset, n=n, cutoff=0.3)

    return [(m, mtalk.encode(m)) for m in matches]

def valid_morse(s: str) -> bool:
    """Check if Morse string only contains valid symbols"""
    parts = s.strip().split()
    return all(p in MORSE_DICT for p in parts)

def morse_analyzer(s: str) -> str:
    # Case 1: Looks like Morse
    if all(ch in ".- " for ch in s.strip()):
        if valid_morse(s):
            decoded = mtalk.decode(s)
            if decoded in DATASET:
                return f" Decoded English: {decoded}"
            else:
                suggestions = recommend_words(decoded, DATASET)
                return f" Word '{decoded}' not in dataset.\nMorse Input: {s}\nSuggestions: {suggestions}"
        else:
            suggestions = recommend_words(s, DATASET)
            return f"Invalid Morse Code.\nInput: {s}\nSuggestions: {suggestions}"

    # Case 2: Looks like English
    else:
        word = s.upper()
        if word in DATASET:
            return f" Morse Code: {mtalk.encode(word)}"
        else:
            encoded = mtalk.encode(word)  # Still give Morse for the entered word
            suggestions = recommend_words(word, DATASET)
            return f"Word '{word}' not in dataset.\nMorse Code: {encoded}\nSuggestions: {suggestions}"


# --- Examples ---
print(morse_analyzer("HELLO"))                 # Valid English
print(morse_analyzer("HILLO"))                 # Invalid English → Suggestions
print(morse_analyzer("....   .   .-..   .-..   ---"))  # Valid Morse
print(morse_analyzer("....   ..   .-..   .-..   ---")) # Invalid Morse → Suggestions
```
